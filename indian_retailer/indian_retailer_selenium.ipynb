```python
import os
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# ---------- CONFIGURATION ----------
download_folder = "/home/justdial/Documents/indianretailer"
os.makedirs(download_folder, exist_ok=True)
file_path = os.path.join(download_folder, "sitemap_news_page.html")

options = webdriver.ChromeOptions()
# options.add_argument("--headless")  # Don't use headless when solving captchas manually
driver = webdriver.Chrome(options=options)

driver.get("https://www.indianretailer.com/sitemap_news_page.xml")

try:
    # --- STEP 1: WAIT FOR IFRAME AND SWITCH TO IT ---
    iframe = WebDriverWait(driver, 15).until(
        EC.presence_of_element_located((By.TAG_NAME, "iframe"))
    )
    driver.switch_to.frame(iframe)
    print("Switched to captcha iframe.")

    # --- STEP 2: CLICK THE CHECKBOX INSIDE IFRAME ---
    checkbox = WebDriverWait(driver, 15).until(
        EC.element_to_be_clickable((By.XPATH, "/html/body/div[1]/div/div[1]/div/div//iframe"))
    )
    driver.execute_script("arguments[0].click();", checkbox)
    print("Checkbox clicked!")

    # Switch back to main page
    driver.switch_to.default_content()

    # --- STEP 3: WAIT UNTIL THE REAL SITEMAP (<urlset>) IS LOADED ---
    WebDriverWait(driver, 60).until(
        EC.presence_of_element_located((By.TAG_NAME, "urlset"))
    )
    print("Sitemap loaded successfully!")

except Exception as e:
    print(f"Error handling captcha or loading sitemap: {e}")

# Save page source
with open(file_path, "w", encoding="utf-8") as f:
    f.write(driver.page_source)

print(f"Sitemap saved to: {file_path}")

driver.quit()
```
